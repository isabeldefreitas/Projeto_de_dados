```python
import pandas as pd
import re
import os
```

```python
#df_certificacoes = pd.read_csv('../ArquivosExternos/dados_brutos/dados_colaboradores 1(Certificacoes).csv') 
```

```python
# variável de ambiente para construir o caminho
caminho_base = os.environ.get("CAMINHO_ARQUIVOS_CSV")
caminho_completo = os.path.join(caminho_base, "dados_brutos", "dados_colaboradores 1(DadosCertificações).csv")

df_certificacoes = pd.read_csv(caminho_completo)
```

```python
df_certificacoes.info()
```

```python
df_certificacoes.rename(columns={'ID': 'idColaborador'}, inplace=True)
```

```python
# Remove espaços antes/depois dos nomes das colunas e normaliza caracteres especiais
df_certificacoes.columns = df_certificacoes.columns.str.strip()

# Reatribui os nomes corretamente via posição
df_certificacoes.columns.values[1] = 'Certificacao_1'
df_certificacoes.columns.values[2] = 'Data_1'
df_certificacoes.columns.values[3] = 'Certificacao_2'
df_certificacoes.columns.values[4] = 'Data_2'
df_certificacoes.columns.values[5] = 'Certificacao_3'
df_certificacoes.columns.values[6] = 'Data_3'
```

```python
display(df_certificacoes)
```

```python
# Tabela 1
tabela1 = df_certificacoes[['idColaborador', 'Certificacao_1', 'Data_1']].dropna(subset=['Certificacao_1', 'Data_1'])
tabela1.columns = ['idColaborador', 'Certificacao', 'Data']

# Tabela 2
tabela2 = df_certificacoes[['idColaborador', 'Certificacao_2', 'Data_2']].dropna(subset=['Certificacao_2', 'Data_2'])
tabela2.columns = ['idColaborador', 'Certificacao', 'Data']

# Tabela 3
tabela3 = df_certificacoes[['idColaborador', 'Certificacao_3', 'Data_3']].dropna(subset=['Certificacao_3', 'Data_3'])
tabela3.columns = ['idColaborador', 'Certificacao', 'Data']
```

```python
display(tabela1)
```

```python
display(tabela2)
```

```python
display(tabela3)
```

```python
tabela_final = pd.concat([tabela1, tabela2, tabela3], ignore_index=True)
display(tabela_final)
```

```python
contagem_certificacoes = tabela_final['Certificacao'].value_counts()
print(contagem_certificacoes)
```

```python
tabela_final.info()
```

```python
def limpar_e_converter(data):
    data = re.sub(r"[^0-9]+", "-", data)

    data = data.strip("-")

    return pd.to_datetime(data, dayfirst=True, errors="coerce")

tabela_final["Data"] = tabela_final["Data"].apply(limpar_e_converter)

# tabela_final["Data de Nascimento"].dt.strftime("%d-%m-%Y")
```

```python
tabela_final.info()
```

```python
# Dicionário de mapeamento (você pode ir completando com os nomes)
traducao = {
    "PowerBI Data Analyst Associate": "Analista de Dados Associado de Power BI",
    "Azure Data Fundamentals": "Conceitos básicos de dados do Azure",
    "Administrator Expert": "Especialista em Administrador",
    # adicione mais conforme necessário
}

# Criar nova coluna traduzida com base no dicionário
tabela_final["Certificacao"] = tabela_final["Certificacao"].replace(traducao)

tabela_final.head(50)
```

```python
#variável de ambiente para construir o caminho
caminho_base_arquivos = os.environ.get("CAMINHO_ARQUIVOS_CSV")
caminho_saida = os.path.join(caminho_base_arquivos, "dados_tratados_v2", "dc_Certificacao_tratado.csv")

# variável caminho_saida
tabela_final.to_csv(caminho_saida, index=False)
```
