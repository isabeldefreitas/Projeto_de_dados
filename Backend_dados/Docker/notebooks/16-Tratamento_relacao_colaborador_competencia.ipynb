```python
# Colaboradores que tinham a competência 10
df_comp10 = df[df['Id_Competencia'] == 10]          
colabs = df_comp10['idColaborador'].dropna().unique()
print(colabs)
```

```python
#Soft skills
softskills = [3, 19]

# Produto cartesiano (colaborador x softskill)
novas_relacoes = (
    pd.MultiIndex
      .from_product([colabs, softskills], names=['idColaborador', 'idSoft_Skills'])
      .to_frame(index=False)
)

novas_relacoes.head(35)
```

```python
# Anexar e garantir unicidade
df_softskills_atualizado = (
    pd.concat([df_fato_softskills, novas_relacoes], ignore_index=True)
)

df_softskills_atualizado.tail(35)
```
