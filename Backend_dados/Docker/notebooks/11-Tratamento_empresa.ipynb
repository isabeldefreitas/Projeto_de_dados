```python
#tratando nome das empresas
df_empresa["nome_empresa"] = df_empresa["nome_empresa"].str.replace("Empresa", "", case=False)
df_empresa.head(10)
```

```python
#padronizar setor da empresa

def padronizar_setor(valor):
    if pd.isna(valor):
        return valor
    valor = valor.strip()

    if len(valor) == 2:
        return valor.upper()
    else:
        # Aplica o title
        valor = valor.title()

        # Lista de preposições que devem ficar em minúsculo
        preposicoes = ["De", "Da", "Do", "Das", "Dos", "Em", "Por", "Sob", "A", "À", "E"]

        # Corrige preposições
        palavras = valor.split()
        palavras_corrigidas = [palavra.lower() if palavra in preposicoes else palavra for palavra in palavras]

        # Garante que a primeira palavra continue com inicial maiúscula
        if palavras_corrigidas:
            palavras_corrigidas[0] = palavras_corrigidas[0].capitalize()

        return ' '.join(palavras_corrigidas)

df_empresa["setor_empresa"] = df_empresa["setor_empresa"].apply(padronizar_setor)
df_empresa.head(10)
```

```python
#tratamento de endereço
# Separar a coluna 'endereco_empresa' em 'logradouro', 'numero' e 'estado'
df_empresa[['logradouro', 'numero', 'estado']] = df_empresa['endereco_empresa'].str.extract(r'^(.*?),\s*(\d+),\s*([A-Z]{2})$')

# Exibir as novas colunas
print(df_empresa[['endereco_empresa', 'logradouro', 'numero', 'estado']])

```

```python
#validação cnpj

import re #importação da biblioteca re para facilitar validação

def validar_cnpj_com_motivo(cnpj):
    cnpj = re.sub(r'\D', '', str(cnpj))  # Remove tudo que não for número

    if len(cnpj) != 14:
        return "Inválido: tamanho incorreto"

    if len(set(cnpj)) == 1:
        return "Inválido: dígitos repetidos"

    return "Válido"
```

```python
df_empresa['cnpj_empresa'].apply(validar_cnpj_com_motivo).value_counts()
```

```python
df_empresa['cnpj_motivo'] = df_empresa['cnpj_empresa'].apply(validar_cnpj_com_motivo)
```
