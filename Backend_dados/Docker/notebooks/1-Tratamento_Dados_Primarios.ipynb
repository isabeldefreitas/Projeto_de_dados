```python
def limpar_e_converter(data):
    data = re.sub(r"[^0-9]+", "-", data)

    data = data.strip("-")

    return pd.to_datetime(data, dayfirst=True, errors="coerce")

df["Data de Nascimento"] = df["Data de Nascimento"].apply(limpar_e_converter)

df["Data de Nascimento"] = df["Data de Nascimento"].dt.strftime("%d-%m-%Y")
```

```python
# 1 - Criar coluna com CPF numérico
df['CPF_Limpo'] = df['CPF'].apply(lambda x: re.sub(r'\D', '', str(x)))

# 2 - Função de validação (somente tamanho e repetição)
def validar_cpf_basico(cpf_num):
    if len(cpf_num) != 11:
        return f"Inválido: possui {len(cpf_num)} dígitos (deve ter 11)"
    if cpf_num == cpf_num[0] * 11:
        return "Inválido: todos os dígitos são iguais"
    return "Válido"

# 3 - Aplicar função
df['Situacao_CPF'] = df['CPF_Limpo'].apply(validar_cpf_basico)
df.drop('CPF_Limpo', axis=1, inplace=True)

print(df)
```

```python
def converter_com_formatos(data):
    formatos = ['%Y-%m-%d', '%d-%m-%Y']
    for fmt in formatos:
        try:
            return pd.to_datetime(data, format=fmt, errors='raise')
        except (ValueError, TypeError):
            continue
    return pd.NaT

df["Data de Nascimento"].apply(converter_com_formatos).isnull().sum()
```
