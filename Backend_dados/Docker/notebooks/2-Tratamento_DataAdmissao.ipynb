```python
import pandas as pd
import re
import os
```

```python
#df = pd.read_csv("../ArquivosExternos/dados_brutos/dados_colaboradores 1(DataAdmissao).csv")
```

```python
# variável de ambiente para construir o caminho
caminho_base = os.environ.get("CAMINHO_ARQUIVOS_CSV")
caminho_completo = os.path.join(caminho_base, "dados_brutos", "dados_colaboradores 1(DataAdmissao).csv")

df = pd.read_csv(caminho_completo)
```

```python
df.info()
```

```python
df = df[['Id', 'Data']]
display(df.head())
```

```python
df.head()
```

```python
def limpar_e_converter(data):
    data = re.sub(r"[^0-9]+", "-", data)

    data = data.strip("-")

    return pd.to_datetime(data, dayfirst=True, errors="coerce")

df["Data"] = df["Data"].apply(limpar_e_converter)
```

```python
df.info()
```

```python
df.head()
```

```python


#variável de ambiente para construir o caminho
caminho_base_arquivos = os.environ.get("CAMINHO_ARQUIVOS_CSV")
caminho_saida = os.path.join(caminho_base_arquivos, "dados_tratados_v2", "dc_DataAdmissao_tratado.csv")

#  variável caminho_saida
df.to_csv(caminho_saida, index=False)
```
