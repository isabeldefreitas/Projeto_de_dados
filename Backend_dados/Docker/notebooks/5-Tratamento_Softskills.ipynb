```python
import pandas as pd
import os
```

```python
# variável de ambiente para construir o caminho
caminho_base = os.environ.get("CAMINHO_ARQUIVOS_CSV")
caminho_completo = os.path.join(caminho_base, "dados_brutos", "dados_colaboradores 1(SoftSkills).csv")

df_soft_skills = pd.read_csv(caminho_completo)
```

```python
# Visualizar as primeiras linhas
df_soft_skills.head(10)
```

```python
print(df_soft_skills.columns)
```

```python
df_soft_skills.rename(columns={'ID': 'idColaborador'}, inplace=True)
```

```python
df_soft_skills_trat = df_soft_skills.melt(
    id_vars='idColaborador',
    value_vars=['Soft Skill 1', 'Soft Skill 2', 'Soft Skill 3', 'Soft Skill 4',
       'Soft Skill 5'],
    value_name='Soft Skills'
)
```

```python
# Remover a coluna gerada com os nomes soft skills 1,2,3 ...
df_soft_skills_trat = df_soft_skills_trat[['idColaborador', 'Soft Skills']]
```

```python
# Visualizar o resultado
display(df_soft_skills_trat.head(10))
```

```python
# Reordenar pelo ID
df_soft_skills_trat = df_soft_skills_trat.sort_values(by='idColaborador').reset_index(drop=True)
display (df_soft_skills_trat)
```

```python
# Verificar presença de NaN (116 vazios)
df_soft_skills_trat.isnull().sum().sum()
```

```python
# Remover NaN (aqui é possível pois trata-se apenas de colaboradores que possuem menos de 5 softskills)
df_soft_skills_trat = df_soft_skills_trat.dropna(subset=['Soft Skills'])
display (df_soft_skills_trat)
```

```python
# Contar as ocorrências de cada soft skill
contagem_soft_skills = df_soft_skills_trat['Soft Skills'].value_counts()

# Criar um df com a contagem
df_contagem = contagem_soft_skills.reset_index()
df_contagem.columns = ['Soft Skill', 'Frequência']

# Contar o número total de categorias únicas
total_categorias = df_contagem.shape[0]

display(df_contagem)
print(f"\nTotal de Soft Skills listadas: {total_categorias}")
```
