```python
# variável de ambiente para construir o caminho
caminho_base = os.environ.get("CAMINHO_ARQUIVOS_CSV")
caminho_completo = os.path.join(caminho_base, "dados_brutos", "dados_colaboradores 1(DadosPCD).csv")

df = pd.read_csv(caminho_completo)
```

```python
df.info()
```

```python
df["Situação PCD"].value_counts()
```

```python
df["Situação PCD"] = df["Situação PCD"].replace("Não", "n")
df["Situação PCD"] = df["Situação PCD"].replace("nao", "n")
df["Situação PCD"] = df["Situação PCD"].replace("NAO", "n")
df["Situação PCD"] = df["Situação PCD"].replace("n", "Não")

df["Situação PCD"] = df["Situação PCD"].replace("sim", "s")
df["Situação PCD"] = df["Situação PCD"].replace("SIM", "s")
df["Situação PCD"] = df["Situação PCD"].replace("s", "Sim")
```

```python
df["Situação PCD"].value_counts()
```

```python
df.rename(columns={"Situação PCD": "Situacao_PCD",
                    "ID" : "idColaborador"}, inplace=True)
```
