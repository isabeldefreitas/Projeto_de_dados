```python

# variável de ambiente para construir o caminho
caminho_base = os.environ.get("CAMINHO_ARQUIVOS_CSV")
caminho_completo = os.path.join(caminho_base, "dados_tratados_v2", "dc_DataAdmissao_Tratado.csv")

df_admissao = pd.read_csv(caminho_completo)
```

```python
# Renomeia as colunas para melhor clareza
df_admissao.rename(columns={'Id': 'idColaborador', 'Data': 'Data_Admissao'}, inplace=True)
```

```python
# Converte a coluna 'Data_Admissao' para o formato de data e hora
df_admissao['Data_Admissao'] = pd.to_datetime(df_admissao['Data_Admissao'])
```

```python

# --- Etapa 1: Criar a Tabela de Dimensão de Data de Admissão ---

# Pega as datas de admissão únicas
dim_data_admissao = pd.DataFrame(df_admissao['Data_Admissao'].unique(), columns=['Data_Admissao'])
# Atribui um ID único a cada data, começando em 1
dim_data_admissao.insert(0, 'idData_Admissao', range(1, 1 + len(dim_data_admissao)))
```

```python
# --- Etapa 2: Modificar a Tabela de Colaboradores para ter a chave da data ---

# Faz um 'merge' da tabela de admissão com a tabela de dimensão de datas para adicionar o 'idData_Admissao'
df_admissao = df_admissao.merge(dim_data_admissao, on='Data_Admissao', how='left')

# faz um 'merge' da tabela de dados primários com a tabela de admissão
dim_colaborador = df_primarios.merge(df_admissao[['idColaborador', 'idData_Admissao']], on='idColaborador', how='left')
```
