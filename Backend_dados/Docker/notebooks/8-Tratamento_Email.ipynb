```python
import os

# variável de ambiente para construir o caminho
caminho_base = os.environ.get("CAMINHO_ARQUIVOS_CSV")
caminho_completo = os.path.join(caminho_base, "dados_tratados_v2", "dc_DadosPrimarios_tratado.csv")

df = pd.read_csv(caminho_completo)
```

```python
def corrigir_email(email):
    if pd.isna(email):  # Caso seja NaN
        return email
    email = str(email).strip()
    
    if email.lower() == "nao informado":
        return email
    
    if "@" not in email:
        return email + "@dominio.com"
    
    usuario, dominio = email.split("@", 1)
    if "." not in dominio:
        dominio += ".com"
    
    return f"{usuario}@{dominio}"

# Aplica a função
df["Email"] = df["Email"].apply(corrigir_email)
```
