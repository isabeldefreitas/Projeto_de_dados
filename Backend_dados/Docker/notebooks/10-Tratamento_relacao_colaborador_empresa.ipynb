```python
# Adiciona o diretório-raiz do projeto ao sys.path
sys.path.append('/app')

import importlib
import utils.conexao
importlib.reload(utils.conexao)
# Importa a função e a variável do schema do módulo de conexão
from utils.conexao import get_dados_origem, SCHEMA_ORIGEM

# consulta 
query = f'''
SELECT 
    c.id_colaborador, 
    e.id_empresa
FROM "{SCHEMA_ORIGEM}"."colaborador" c
LEFT JOIN "{SCHEMA_ORIGEM}"."colaborador_empresa" ce 
    ON c.id_colaborador = ce.id_colaborador
LEFT JOIN "{SCHEMA_ORIGEM}"."empresa" e 
    ON ce.id_empresa = e.id_empresa;
'''
df = get_dados_origem(query) 

if df is not None:
    print("Extração concluída:")
    print(df.head())
```
